### 1. Relation Plot
- to see the statistical relation between 2 or more variables.
- Bicariate Analysis

Plots under this section
- Scatterplot
- Lineplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
# scatter plot -> axes level function
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
# relplot -> figure level -> square shape
sns.relplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size',kind='scatter')
```

```python
# scatter using relplot -> size and hue
```

```python
# style sementics
```

```python
# line plot
gap = px.data.gapminder()
temp_df = gap[gap['country']=='India']
temp_df
```

```python
# axes level function
sns.lineplot(data=temp_df,x='year',y='lifeExp')
```

```python
# using relplot
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line')
```

```python
# hue -> style
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line',hue='country',style='continent',size='continent')
```

```python
sns.lineplot(data=temp_df,x='year',y='lifeExp',hue='country',style='continent',size='continent')
```

```python
# facet Plot -> figure level fuction -> work with relplot
# it will not work with scatteplot and lineplot

sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',col='sex',row='day')
```

```python
# col wrap
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=2)
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',col='sex',row='day')
```

### 2. Distribution Plots
- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers

Plots under distribution plot
- histplot
- kdeplot
- rugplot


```python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
```

```python
# Plotting univariate histogram
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
# bins parameter
sns.displot(data=tips,x='total_bill',kind='hist',bins=5)
```

```python
# It's also possible to visualize the distribution of a categorical using the logic of a histogram.
# Discrete bins are automatically set for categorical variables

# countplot
sns.displot(data=tips,x='day',kind='hist')
```

```python
# hue parameter
sns.displot(data=tips,x='tip',kind='hist',hue='sex')
```

```python
# element -> step
sns.displot(data=tips,x='tip',kind='hist',hue='sex',element='step')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic,x='age',kind='hist',hue='sex',element='step')
```

```python
# faceting using col and row -> not work on histplot function
sns.displot(data=tips,x='tip',kind='hist',col='sex',element='step')
```

```python
# kdeplot
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a contnuous density esimate
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.displot(tips,x='total_bill',kind='kde')
```

```python
# hue -> fill
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=5,aspect=2)
```

```python
# Rugplot
# Plot marginal distribution by drawing tick along the x and y axes.
# This function is intended to complement other plots by showing the location of indicidual observation in unobtrusive way.

sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill',height=0.2)
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observations within each rectangle with the fill color

# sns.histplot(data=tips,x='total_bill',y='tip')
sns.displot(data=tips,x='total_bill',y='tip',kind='hist')
```

```python
# Bivariate Kdeplot
# A bivariate KDEplot smoothes the (x,y) observations with 2D Gaussian
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

### 2. Matrix Plot
- Heatmap
- Clustermap

```python
# Heatmap
# Plot rectangular data as a color-encoded matrix
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')

# axes level function
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
# annot
temp_df = gap[gap['continent']=='Europe'].pivot(index='country',columns='year',values='lifeExp')

plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidth=0.5,cmap='summer')
```
