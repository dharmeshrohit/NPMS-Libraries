### Timestamp Object
Time stamps reference particular moments in tim (e.g., Oct 15th, 2025 at 7:00pm)

### Creating Timestamp objects

```python
# creating a timestamp
type(pd.Timestamp('2023/1/5'))
```

```python
# Variations
pd.Timestamp('2023-1-5')
pd.Timestamp('2023, 1, 5')
```

```python
# only year
pd.Timestamp('2023')
```

```python
# using text
pd.Timestamp('5th January 2023')
```

```python
# providing time also
pd.Timestamp('5th January 2023 9:21AM')
```

```python
# using datatime.datetime object
import datetime as dt

x = pd.Timestamp(dt.datetime(2023, 1, 5, 9, 21, 56))
x
```

```python
# fetching attributes
x.year
x.month
x.day
x.hour
x.minute
x.second
```

```python
# why seperate objects to handle data adn time when python already has datetime functionality?
```

- Syntax wise datetime is very convenient
- But the performance takes a hit while working with huge data. List and Numpy Array
- The weakness of Python's datetime format inspired the Numpy team to add a set of native time series data type to Numpy.
- The datetime64 dtype encodes dates as 64-bit integers, and thus allows arrays of dates to be represented very compactly.

```python
import numpy as np
date = np.array('2015-07-04', dtype=np.datetime64)
date
```

```python
date + np.arange(12)
```

- Because of the uniform type in Numpy datetim64 array, this type of operatio can be accomplished much more quickly than if we were working directly with Python's datetime objects, especially as arrays get large

- Pandas Timestamp object combine the ease-of-use of python datetime with the efficient storage and vectorized interface of numpy.datetime64

- From a group of these Timestamp objects, Pandas can contruct a DatatimeIndex that can be used to index data in a Series or DateFrame

### DatetimeIndex Object
A collection of pandas timestamp

```python
# from strings
type(pd.DatetimeIndex(['2023/1/1', '2022/1/1', '2021/1/1']))
```

```python
# using python datetime object
pd.DatetimeIndex([dt.datetime(2023, 1, 1),dt.datetime(2022,1,1),dt.datetime(2021,1,1)])
```
